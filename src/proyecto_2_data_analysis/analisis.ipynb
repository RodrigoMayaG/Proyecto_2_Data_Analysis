```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/Users/rodrigomayagonzalez/Documents/Python Proyectos aprender/Proyecto_2_Data_Analysis/archive/DataCoSupplyChainDataset.csv', encoding='latin-1')
df
```

```python
data_limpia = df[['Type', 'Days for shipping (real)', 'Days for shipment (scheduled)', 'Delivery Status', 'Late_delivery_risk',
                  'Category Name', 'Customer City', 'Customer Country', 'Order Country', 'order date (DateOrders)',
                  'Order Id', 'Sales per customer', 'Order Item Product Price', 'Order Item Quantity', 'Order Item Total',
                  'Order Status', 'shipping date (DateOrders)', 'Shipping Mode']]
data_limpia
```

```python
data_order = data_limpia[['Order Country', 'Order Id', 'Order Status']]
data_order
```

```python
data_order_group = data_order.groupby(['Order Country', 'Order Status'])['Order Id'].nunique().unstack(fill_value=0)
data_order_group.sort_values('COMPLETE', ascending= False)
data_order_group['Total'] = data_order_group.sum(axis=1)
data_order_group.sort_values(by= 'Total', ascending=False)
```

```python
pct = (data_order_group.drop(columns= 'Total')
       .div(data_order_group['Total'], axis= 0)
       .mul(100).round(1))
pct['Total_Orders'] = data_order_group['Total']
pct.sort_values(by='Total_Orders', ascending=False)
```

```python
data_status = data_limpia[data_limpia['Order Status'] != 'COMPLETE']
data_status
```

```python
data_status_group = data_status.groupby('Order Status')['Order Id'].nunique().reset_index()
total = data_status_group['Order Id'].sum()
data_status_group
```

```python
data_status_group['%de Total'] = round(data_status_group['Order Id'] / total * 100, 2)
data_status_group = data_status_group.sort_values(by='%de Total', ascending= False)
data_status_group
```

```python
plt.bar(data_status_group['Order Status'], data_status_group['%de Total'])
plt.xticks(rotation= 90)
plt.show()
```

```python
data_status_group_delivery = data_status.groupby('Delivery Status')['Order Id'].nunique().reset_index()
data_status_group_delivery = data_status_group_delivery.sort_values(by= 'Order Id', ascending=False)
data_status_group_delivery
```
